```python
import pandas as pd
import numpy as np
import ast
```

```python
movies = pd.read_csv(r'dataset.csv')
```

```python
movies
```

```python
movies.isna().sum()
```

```python
movies.dropna(ignore_index = True , inplace = True)
```

```python
movies
```

```python
movies['overview'][9981]
```

```python
movies['title'][99]
```

```python
movies['tags'] = movies['genre'] + movies['overview']
```

```python
movies.head()
```

```python
df = movies[['id' , 'title' , 'tags']].copy()
```

```python
df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features = 10000 , stop_words = 'english')
```

```python
X = cv.fit_transform(df['tags']).toarray()
```

```python
cv
```

```python
X.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sim = cosine_similarity(X)
```

```python
sim
```

```python
dist = sorted(list(enumerate(sim[0])), key=lambda X: X[1], reverse=True)
#returns a list as (index , similarity) as 1 element
#sim[0] means similarity to the movie at index 0
```

```python
dist
```

```python
for i in dist[0:5]:
    print(df.iloc[i[0]].title)
```

```python
def recommend(movie):
    index = df[df['title'] == movie].index[0]
    rec = []
    _dist = sorted(list(enumerate(sim[index])), reverse=True, key=lambda X: X[1])
    for i in _dist[1:6]:
        rec.append(df.iloc[i[0]].title)
    return rec
```

```python
recommend('Dilwale Dulhania Le Jayenge')
```

```python
import pickle
# Save the similarity matrix
with open('similarity.pkl', 'wb') as f:
    pickle.dump(sim, f)

# Save the dataframe 
with open('movies.pkl', 'wb') as f:
    pickle.dump(df, f)

#Save the vectorizer 
with open('vectorizer.pkl', 'wb') as f:
    pickle.dump(cv, f)
```
